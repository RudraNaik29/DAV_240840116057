**Step - 1 Import Pandas library**

```python
import pandas as pd                                 # Import Pandas Library
```

**Step - 2 Create a dataset**

```python
data = {
    "ID": [ 101, 102, 103, 104, 105, 106, 107],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krisha", "Harry"],                               #Create Data
    "Age": [20, 21, 19, 20, 22, 24, 28],
    "City": ["Surat", "Navsari", "Valsad", "Bharuch", "Mumbai", "Bangalore", "Pune"],
     "Marks": [20, 90, 15, 50, 55, 65, 60]
}
```

**Step - 3 Create and display dataset**

```python
df = pd.DataFrame(data)                # Convert into DataFrame
print(df)                              # To Print The DataFrame
```

**Step - 4 Save dataset as csv file**

```python
df.to_csv("Students.csv", index=False)                     #Convert data into a CSV File


print("CSV file is created")
```

**Step - 5 Save Dataframe As Excel**

```python
df.to_excel("Students.xlsx", index=False)                     #Convert data into a EXCEL File


print("Excel file is created")
```

**Step - 6 Save dataframe As Json**

```python
df.to_json("Students.json", index=False)                  #Convert into jSON File

print("JSON file is created")
```
