**step - 1 : import Pandas Library**

```python
import pandas as pd               # Import Pandas
```

**step-2:Create dataframe**

```python
data = {
    "ID" : [101,102,103,104,104, 105, 106, 107],
    "Name" : ["Rahul", "Priya", "Amit", None, "Riya", "Ajay", "Naresh", "Jenil"],
    "Age" :   [20, 21, None, 20, 20, 22, 23, 28],
    "City" : ["Surat", "Navsari", "Surat", "Bharuch", "Bharuch", "Navsari", "Surat", "Ahemdabad"],
    "Marks" : [81, None, 78, 88, 88, 81, 90, 93]
}

df = pd.DataFrame(data)
print(df)
```

**step 4:check missing values**

```python
df.isnull()                               # Display Missing Values
```

```python
df.isnull().sum()                          # Count Null Values
```

```python
df.isnull().sum().sum()                          # Count NullValues for entire Dataset
```

**step 5:fill missing values**

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())         # Fill Missing Values

df
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())                # Fill Missing Value in Age Column

df
```

```python
df["Name"] = df["Name"].fillna("Rudra")           # Fill Missing Value in Name

df
```

**step 6:Detect duplicate rercords**

```python
df.duplicated()                         # Check Duplicate Value
```

```python
df.duplicated().sum()                 # Count Duplicated Rows
```

**step 7: remove the dupolicate records**

```python
df.drop_duplicates()                                       # Remove Duplicated Records
```

**step 8:Rename the column.**

```python
df.rename(columns={"Marks" : "Score"}, inplace=True)           # Rename Columns
df
```

**step 9:Change data type**

```python
df["Age"] = df["Age"].astype(int)          #Convert the age column from float to integer
df
```

**step 10:Create a new column**

```python
df["Final_Score"] = df["Score"] + 5           #Create a new column.
df
```

**Step - 11 Convert text in uppercase**

```python
df["Name"] = df["Name"].str.upper()                 # Convert Into UpperCase
df
```

**Step - 12 Convert text into lower case**

```python
df["City"] = df["City"].str.lower()                  # Convert Text into LowerCase
df
```

**Step - 13 Remove extra space**

```python
df["City"] = df["City"].str.strip()             # Remove Extra Space
df
```

**Step - 14 sort the dataset**

```python
df.sort_values(by=["Final_Score"], inplace=True)              # Sort The DataSet Values
df
```

**Step - 15 Save the clean data**

```python
df.to_csv('Cleaned_Students.csv', index=False)              # Save Cleaned DataSet
df
```
