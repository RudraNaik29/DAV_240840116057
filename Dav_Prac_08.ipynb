Step-1 Import Libraries

```python
# Import Libraries.
import pandas as pd    # Import Pandas Library.

import numpy as np     # Import NumPy Library.

import matplotlib.pyplot as plt    # Import Matplot Library.
```

**Step-2 Create A DataSet**

```python
  # Create Dataset.
data = {
    "Month": ["Jan", "Feb", "March", "April", "May" , "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"],
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West", "North", "South", "East", "West"],
    "Sales": [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold": [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40],
    "Profit" : [18000, 27000, 30000, 22000, 32000, 35000, 28000, 34000, 38000, 33000, 46000, 50000]
}
df = pd.DataFrame(data)
df
```

**Step-3 Create An Interactive Line Chart**

```python
# Line Chart Based on Revenue Change Throughout The Year
import plotly.express as px
fig = px.line(
    df,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Sales Trend"
)
fig.show()
```

**Step-4 Improve The Hover Information**

```python
#Create a Line Chart using Sales over Month
import plotly.express as px
fig = px.line(df, x='Month', y='Sales', title='Sales Over Month', markers=True)
fig.show()
```

```python
# Add additional Column to the hover information
import plotly.express as px
fig = px.line(
    df,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Sales Trend with Extra Hover Details",
    hover_data=["Profit", "Units_Sold", "Region"]
)
fig.show()
```

**Step-5 Create an Interactive Bar Chart**

```python
import plotly.express as px

# Grouping and calculating total sales
product_sales = df.groupby('Product')['Sales'].sum().reset_index()

# Creating the interactive bar chart
fig = px.bar(
    product_sales,
    x='Product',
    y='Sales',
    title='Total Sales by Product',
    labels={'Sales': 'Total Sales ($)', 'Product': 'Product Name'},
    text_auto='.2s'
)
fig.show()
```

**Step-6 Create an Interactive Scatter Plot**

```python
# To Analyze the relationship between Units Sold and Sales using Interactive Scatter Plot
import plotly.express as px

fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    color="Product",
    size="Profit",
    hover_data=["Month", "Region"],
    title="Relationship Between Units Sold and Sales"
)
fig.show()
```

```python
# Interactive Scatter Plot with Color mapping
import plotly.express as px

fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    color="Region",
    size="Profit",
    hover_data=["Month", "Product"],
    title="Units Sold vs Sales by Region"
)
fig.show()
```

**Step 7 :- Create an interactive Pie Chart**

```python
import plotly.express as px

region_sales = df.groupby('Region')['Sales'].sum().reset_index()

# Create an interactive Pie Chart to visualize the Sales distribution by Region
fig = px.pie(
    region_sales,
    values='Sales',
    names='Region',
    title='Sales Distribution by Region',
    hole=0.3
)

# Display the interactive pie chart
fig.show()
```

**Step 8 :- Plotly Graph Objects**

```python
# Import Plotly Graph Objects
import plotly.graph_objects as go

# Initialize the Figure
fig = go.Figure()

# Add the trace using go.Scatter for line and marker plots
fig.add_trace(go.Scatter(
    x=df['Month'],
    y=df['Sales'],
    mode='lines+markers',
    name='Sales'
))

# Display the figure
fig.show()
```
