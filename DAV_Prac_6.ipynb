**Aim - Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie)**

**Step-1 Import Libraries**

```python
# Import Libraries.
import pandas as pd    # Import Pandas Library.

import numpy as np     # Import NumPy Library.

import matplotlib.pyplot as plt    # Import Matplot Library.
```

**Step-2 Create a Dataset**

```python
  # Create Dataset.
data = {
    "Month": ["Jan", "Feb", "March", "Arpil", "May" , "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"],
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West", "North", "South", "East", "West"],
    "Sales": [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold": [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
df = pd.DataFrame(data)
df
```

**Step-3 Create a Line Chart**
**How Did Company Sales Change from Jan to Dec**

```python
#Create A Line Chart based on Company Sale Change from Jan to Dec
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], marker="o", color="blue")
plt.title("Monthly Sales Trend (Jan to Dec)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

**Step-4 Create a Barchart , Which product generate how much sales**

```python
#Create a BarChart Based on Which Product Generate How Much Sales
product_sales = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(7, 5))
plt.bar(product_sales.index, product_sales.values, color=["skyblue", "orange", "green"])
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

**Step-5  Create a histogram , How Sales are Distributed**

```python
#Create a Histogram based on How Sales Are Distributed
plt.figure(figsize=(7, 5))
plt.hist(df["Sales"], bins=5, color="purple", edgecolor="black")
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

**Step -6 Create a Scatter Plot , Is There a Relationship Between The numder of unit sold and total sales**

```python
#Create A Scikit Plot based on Relationship Between the nmber of unit sold and total sales
plt.figure(figsize=(7, 5))
plt.scatter(df["Units_Sold"], df["Sales"], color="red")
plt.title("Units Sold vs Sales")
plt.xlabel("Units Sold")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

**Step-7 Create a Pi Chart , What Percentage of Total Sales Come From Each Region**

```python
# Create a Pi Chart based on Total Sales from Each Region
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(6, 6))
plt.pie(region_sales.values, labels=region_sales.index, autopct="%1.1f%%", startangle=90)
plt.title("Sales Share by Region")
plt.show()
```
